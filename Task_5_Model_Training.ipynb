```python
import pandas as pd
import numpy as np
import joblib

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

print("Libraries imported successfully.")
```

```python
iris = load_iris()

X = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

y = pd.Series(
    iris.target,
    name="target"
)

print("Dataset loaded successfully.")
print("Rows:", X.shape[0])
print("Features:", X.shape[1])

display(X.head())
```

```python
print("Feature names:")
for feature in iris.feature_names:
    print("-", feature)

print("\nTarget classes:")
for index, name in enumerate(iris.target_names):
    print(index, "→", name)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=3,
    random_state=42
)

model.fit(X_train, y_train)

print("Decision Tree trained successfully.")
```

```python
y_pred = model.predict(X_test)

accuracy = accuracy_score(
    y_test,
    y_pred
)

print(f"Accuracy: {accuracy:.4f}")
print(f"Accuracy: {accuracy * 100:.2f}%")
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
print("Classification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=iris.target_names
    )
)
```

```python
model_bundle = {
    "model": model,
    "class_names": iris.target_names.tolist(),
    "feature_names": iris.feature_names
}

print("Model bundle created successfully.")
```

```python
MODEL_FILE = "iris_model.joblib"

joblib.dump(
    model_bundle,
    MODEL_FILE
)

print("Saved:", MODEL_FILE)
```

```python
loaded_bundle = joblib.load("iris_model.joblib")

loaded_model = loaded_bundle["model"]

loaded_predictions = loaded_model.predict(X_test)

loaded_accuracy = accuracy_score(
    y_test,
    loaded_predictions
)

print(f"Original accuracy: {accuracy * 100:.2f}%")
print(f"Loaded model accuracy: {loaded_accuracy * 100:.2f}%")

assert np.array_equal(
    y_pred,
    loaded_predictions
)

print("Model verification successful.")
```
